<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Item в C#, который будет представлять информацию о
товарах, которые могут быть заказаны или возвращены. На основе этого класса
разработать 2-3 производных класса, демонстрирующих принципы наследования и
полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и
методы, а также переопределены некоторые методы базового класса для
демонстрации полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

```
public class Item
{
    public int ItemId { get; set; }
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Item(int id, string name, decimal price)
    {
        ItemId = id;
        Name = name;
        Price = price;
    }

    public virtual string GetDetails()
    {
        return ItemId + ". " + Name + " - " + Price + " руб.";
    }

    public virtual decimal CalculateDiscount()
    {
        return Price * 0.05m;
    }

    public virtual void ApplyDiscount(decimal discount)
    {
        Price -= discount;
        Console.WriteLine("Новая цена со скидкой для " + Name + ": " + Price + " руб.");
    }
}

public class SingleItem : Item
{
    public string UnitMeasure { get; set; }

    public SingleItem(int id, string name, decimal price, string unit) 
        : base(id, name, price)
    {
        UnitMeasure = unit;
    }

    public override string GetDetails()
    {
        return base.GetDetails() + " (Ед. изм.: " + UnitMeasure + ")";
    }
}

public class PackageItem : Item
{
    public int QuantityPerPackage { get; set; }

    public PackageItem(int id, string name, decimal price, int count) 
        : base(id, name, price)
    {
        QuantityPerPackage = count;
    }

    public override decimal CalculateDiscount()
    {
        if (QuantityPerPackage > 5)
            return Price * 0.15m;

        return Price * 0.10m;
    }
}

public class SpecialItem : Item
{
    public DateTime DiscountExpirationDate { get; set; }

    public SpecialItem(int id, string name, decimal price, DateTime expDate) 
        : base(id, name, price)
    {
        DiscountExpirationDate = expDate;
    }

    public override void ApplyDiscount(decimal discount)
    {
        if (DateTime.Now > DiscountExpirationDate)
        {
            Console.WriteLine("Скидка на " + Name + " уже не действует!");
            return;
        }

        base.ApplyDiscount(discount);
        Console.WriteLine("Скидка активна до: " + DiscountExpirationDate.ToShortDateString());
    }
}

public class Order
{
    public int OrderId { get; set; }
    public Item[] Items { get; set; }

    public Order(int id, Item[] items)
    {
        OrderId = id;
        Items = items;
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;

        for (int i = 0; i < Items.Length; i++)
        {
            total += Items[i].Price;
        }

        return total;
    }
}

// Проверка работы
SingleItem milk = new SingleItem(1, "Молоко", 80, "л");
PackageItem juice = new PackageItem(2, "Сок (упаковка)", 600, 6);
SpecialItem cake = new SpecialItem(3, "Торт", 500, DateTime.Now.AddDays(2));

Console.WriteLine(milk.GetDetails());
Console.WriteLine(juice.GetDetails());

Console.WriteLine();

decimal juiceDiscount = juice.CalculateDiscount();
juice.ApplyDiscount(juiceDiscount);
cake.ApplyDiscount(100);

Console.WriteLine();

Item[] myItems = new Item[] { milk, juice, cake };
Order myOrder = new Order(10, myItems);

Console.WriteLine("Итого к оплате по заказу №" + myOrder.OrderId + ": " + myOrder.GetTotalPrice() + " руб.");
```
